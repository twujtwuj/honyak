```python
# Load in a new notebook
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL_PATH = "training/models/t5-honyak-10-epoch"  # big gives translation
tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_PATH)
```

```python
def translate_to_toki_pona(sentence, max_len=40, num_beams=5):
    prompt = f"translate English to Toki Pona: {sentence}"
    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_length=max_len,
        num_beams=num_beams,
        early_stopping=True
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
examples = [
    "Moom has three dogs.",
    "I want to eat food.",
    "This is a test sentence."
]

for ex in examples:
    tp = translate_to_toki_pona(ex)
    print(f"EN: {ex}  -> TP: {tp}")
```

```python
name_examples = [
    "Tom is tall.",
    "Karen went to the store yesterday.",
    "Helen saw Beatrice."
]

for ex in name_examples:
    tp = translate_to_toki_pona(ex)
    print(f"EN: {ex}  -> TP: {tp}")
```

```python
long_examples = [
    "Tom is tall. He went to the store.",
    "Karen went to the store yesterday.",
    "Helen saw Beatrice."
]

for ex in long_examples:
    tp = translate_to_toki_pona(ex)
    print(f"EN: {ex}  -> TP: {tp}")
```
