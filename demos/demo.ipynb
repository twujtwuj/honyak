# Demo

This notebooks is a demonstration of the EN -> TP translation capacities of our fine-tuned model.

```python
# Load in a new notebook
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from honyak.inference_utils import translate_to_toki_pona

MODEL_PATH = "../training/models/t5-honyak-10-epoch_v3"  # big gives translation
tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_PATH)
```

```python
examples = [
    "Moom has three dogs.",
    "I want to eat food.",
    "This is a test sentence.",
    "Tom is fat. Did you know that?",
    "I want you to try harder."
]

for ex in examples:
    tp = translate_to_toki_pona(model, tokenizer, ex)
    print(f"EN: {ex}  -> TP: {tp}")
```

```python
name_examples = [
    "Tom is tall.",
    "Karen went to the store yesterday.",
    "Helen saw Beatrice."
]

for ex in name_examples:
    tp = translate_to_toki_pona(model, tokenizer, ex)
    print(f"EN: {ex}  -> TP: {tp}")
```

```python
long_examples = [
    "Tom is tall. He went to the store.",
    "Karen went to the store yesterday.",
    "Helen saw Beatrice."
]

for ex in long_examples:
    tp = translate_to_toki_pona(model, tokenizer, ex)
    print(f"EN: {ex}  -> TP: {tp}")
```
