# Demo

This notebooks is a demonstration of the EN -> TP translation capacities of our fine-tuned model.

```python
# Load in a new notebook
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from honyak.inference_utils import translate_to_toki_pona
import torch

MODEL_PATH = "../training/models/t5-honyak-10-epoch_v3"  #/checkpoint-1412"  # big gives translation
tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_PATH)
model.eval();
```

#### Short example sentences

The model deals well with translating short, simple, unambiguous sentences that resemble those found in the corpus. The model has a very good grasp of the grammar of the language and consistently gives translations that are word-for-word the same as my own.

```python
examples = [
    # Basic, probably regurgitated sentences
    "The girl has animals.",

    # Context marking using 'la'
    "I want to eat food during the day.",
    
    # Question refactoring
    "The sun is big. Did you know that?",
    "Did you know that the sun is big?",

    # Multiple agents
    "I want you to try more things.",
    "The mother wanted to give him the water."
]

for ex in examples:
    tp = translate_to_toki_pona(model, tokenizer, ex)
    print(f"EN: {ex}  -> TP: {tp}")
```

#### Tokiponification of names

Toki Pona has very strict phonotactics, meaning that only around 100 syllables are legal in this languages and some syllables cannot occur in sequence.

In this version of the model (`t5-honyak-10-epoch_v3`) the model can only tokiponify certain name correctly. These were either found directly in the corpus and repeated multiple times, or they ressemble their tokiponification closely.

In future iterations of the model, I want to augment the corpus with far more examples of tokiponifications. I also want to strictly constrain the output of the model so that it can only produce legal tokiponifications (e.g. by having the model only output TP tokens). Currently, the model does not understand what 'legal' symbols and syllables are. 

```python
name_examples = [

    # In training corpus and seen frequently
    "Tom is tall.",
    "Mary saw Susan.",

    # In training corpus but seen rarely
    "Tokyo is a big city.",
    "Do you know Daniel?",

    # Close to tokiponification
    "Tomi went to the store yesterday.",  # tokiponification would be 'Tomi'
    
    # Not in corpus and not close to Tokiponification
    "Let's go to Bolivia."
]

for ex in name_examples:
    tp = translate_to_toki_pona(model, tokenizer, ex)
    print(f"EN: {ex}  -> TP: {tp}")
```
